#### Step 1: Import libraries and load the environment variables

````python
import os
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import Markdown, display

load_dotenv()


OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")

if OPENAI_API_KEY is None:
    raise Exception("API key is missing")
````

#### Step 2: Call the API

````python
client = OpenAI()
response = client.chat.completions.create(
    model='gpt-4o-mini',
    messages=[
        {
            "role" : "system",
            "content" : "you are helpfull assistance"
        },
        {
            "role" : "user",
            "content" : "why the sky is blue?"
        }
    ],
    max_tokens = 100,
    temperature=0.5
)
print(response.choices[0].message.content)
````

#### Step 3: Customize the User Prompt

Find the best career suited to you

````python
system_message = "You are a career path mentor with 20 plus years of expirence"

user_prompt = """
Help me find my ideal career based on the following:

1. My Skillset
- Java,python, camunda, n8n , little about agentic AI
- Keycloack , springboot,fastAPI
- Tools : intellij , postman, claude, chat gpt, cursor.

2. My current industry
- software developer in bluespacelabs (IT)

3. My interests
- Understading and create the LLM,genAI,agentic AI model from scratch.
- Solving logic problems
- Exploring new tech 

4. My objectives
- To become AI engineer with good salary.
- Need to create a my own product to solve the issues using AI
- Always up to date with the latest update on AI

Now based on this, give me:
- career path
- startup ideas
- best company to join eventually get insights on startup
"""


response = client.chat.completions.create(
    model='gpt-4o-mini',
    messages=[
        {
            "role" : "system",
            "content" : system_message
        },
        {
            "role" : "user",
            "content" : user_prompt
        }

    ]
)
display(Markdown(response.choices[0].message.content))

````

#### Step 4: Build a plan for your career path

````python
# Step 1: Use your chat model result and copy/paste your career suggestion below
career_suggestion = response.choices[0].message.content
# TODO: Paste your career suggestion here

# Step 2: Write a reasoning prompt that will help the model plan how to pursue this career
# Your prompt must:
# - Refer to the user's profile + their chosen career
# - Ask for a step-by-step plan
# - Include at least 3 types of guidance (e.g. skills, resources, milestones)
reasoning_prompt = f"""
I have been following career suggestion based on my profile : {career_suggestion}

here is is my profile again {user_prompt}

Please provide a step by step plan to pursue this career . The plan should include : 
- Key skill i need to develop 
- Resources i can use to learn these skill 
- Milestone to track my progress

"""  
# TODO: Write this yourself


completion = client.chat.completions.create(
    model="gpt-6-luna",
    messages=[
        {"role": "system", "content": system_message},
        {"role": "user", "content": reasoning_prompt}
    ]
) 

reasoning_response = completion.choices[0].message.content
display(Markdown(reasoning_response))
````
